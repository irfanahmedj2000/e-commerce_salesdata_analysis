
During the initial data exploration, three columns were removed from the dataset:
- **Item Quantity**: Every order had a quantity of 1, so the column contained no variation.
- **Order Shipped From (State)**: All orders were shipped from the same state, making the column unsuitable for comparative analysis.
- **Shipping Charges**: Shipping charges were identical for every transaction and therefore did not provide any additional analytical value.

These columns were removed to simplify the dataset while preserving all information relevant to sales analysis.



The **Usual Price** column contained 100% missing values and therefore did not provide any usable information for the analysis. Since the column could not contribute to any business insights or statistical analysis, it was removed from the cleaned dataset.

```python
sales_clean = sales_clean.drop_duplicates()
sales_clean.duplicated().sum()
```



The dataset contained duplicate transaction records.

Upon inspection, the duplicate rows had identical values across all columns, including Order ID, Order Item ID, Product Description, Customer Information, Order Date, and Invoice Amount.

These duplicates appeared to be repeated exports of the same transaction rather than legitimate multiple purchases.

Therefore, duplicate records were removed to ensure accurate sales and revenue analysis.

```python
sales_clean['Order Date'] = pd.to_datetime(sales_clean['Order Date'])

sales_clean['Order Approval Date'] = pd.to_datetime(sales_clean['Order Approval Date'])
```

```python
sales_clean['Year'] = sales_clean['Order Date'].dt.year
sales_clean['Month'] = sales_clean['Order Date'].dt.month_name()
sales_clean['Month_Number'] = sales_clean['Order Date'].dt.month
sales_clean['Day'] = sales_clean['Order Date'].dt.day
sales_clean['Weekday'] = sales_clean['Order Date'].dt.day_name()
```

```python
sales_clean.loc[
    sales_clean['Total Discount'] < 0,
    [
        'Price before discount',
        'Total Discount',
        'Bank Offer Share',
        'Price after discount (Price before discount-Total discount)'
    ]
]
```


The **Total Discount** column contains negative values. After validating the calculations, it was observed that the dataset stores discounts as negative amounts.

The relationship is:

Price After Discount = Price Before Discount + Total Discount

Therefore, the negative values represent discount deductions rather than data quality issues. No transformation was applied to preserve the original business records.
